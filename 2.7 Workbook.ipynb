```python
# Create a copy with fewer columns 

df_1 = df.drop(columns = {'started_at', 'ended_at', 'ride_id', 
'rideable_type', 'start_station_id', 'end_station_id', 'member_casual'})
```

```python
np.random.seed(32)
red = np.random.rand(len(df_1)) <= 0.92
```

```python
small = df_1[~red]
small.shape
```

```python
# Group by start station and calculate the average ride duration
top_20_stations = (
    small.groupby("start_station_name")
    ["ride_duration"].mean()
    .nlargest(20)
    .reset_index()
)

# Display the resulting dataframe
top_20_stations
```

```python
top_20_stations.to_csv(os.path.join(path, 'Data','Prepared_Data', 'RideDuration_top20.csv'),index = False)
```
